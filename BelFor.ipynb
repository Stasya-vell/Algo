```python
n = 5
# Задаем граф списком ребер
graph = [
    (0, 1, -1),
    (0, 2, 4),
    (1, 2, 3),
    (1, 3, 2),
    (3, 1, 1),
    (1, 4, 2),
    (4, 3, -3),
    (3, 2, 5)
]
src = 0
```

```python
def belfor(n, graph, src):
    # Дистанция на нулевом шаге до вершин
    dist = [float('inf')] * n
    dist[src] = 0
    # Чтобы дойти до любой вершины нужно не более n-1 шага
    for i in range(n-1):
        # Перебеираем все ребра, если есть более легкий маршрут то запоминаем его
        for j in graph:
            if dist[j[1]] > dist[j[0]] + j[2]:
                dist[j[1]] = dist[j[0]] + j[2]
    # Если можно еще облегчить путь то есть цикл отрицательного веса
    for j in graph:
        if dist[j[1]] > dist[j[0]] + j[2]:
            return "There is a negative-weight cycle"
    return dist
```

```python
print(belfor(n, graph, src))
```
