```python
import datetime

def  leave(date):
    '''Прибавляем день чтобы узнать дату, когда гостя уже не будет в отеле.'''
    Date = datetime.datetime.strptime(date, "%Y-%m-%d")
    Tomorrow = Date + datetime.timedelta(days=1)
    tomorrow = Tomorrow.strftime("%Y-%m-%d")
    return tomorrow

def hotel(data): #O(n log n), O(n)
    '''Записываем даты приезда и отъезда гостей в словарь, в качестве значений +1 если гость приехал и -1 если уехал.
    Сортируем словарь по ключам. Проходим по словарю для поиска дня с наибольшим числом гостей.'''
    visits = dict()
    for guest in data:
        if guest[0] in visits:
            visits[guest[0]] += 1
        else:
            visits[guest[0]] = 1
        tomorrow = leave(guest[1])
        if tomorrow in visits:
            visits[tomorrow] -= 1
        else:
            visits[tomorrow] = -1
    visits = sorted(visits.items())
    day = 0
    peak = visits[0]
    for visit in visits:
        day += visit[1]
        if day > peak[1]:
            peak = (visit[0], day)
    return peak
```

```python
data = [("2024-09-15", "2024-09-15"), ("2024-09-14", "2024-09-21")]
print(*hotel(data))
```
