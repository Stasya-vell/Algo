```python
def bookmaker(prices, k): #O(n*k), O(k)
    #Создаем списки buy - для купленных акций (потраченные деньги), revenue - для полученной прибыли
    buy = [float('-inf') for _ in range(k)]
    revenue = [0 for _ in range(k)]
    for p in prices:
        for i in range(k):
            #Продолжаем держать акцию или покупаем нувую после i-1 завершенной сделки
            buy[i] = max(buy[i], revenue[i-1] - p)
            #Продолжаем держать акцию или продаем акцию купренную на i сделке
            revenue[i] = max(revenue[i], buy[i] + p)
    #Возвращаем прибыль после k сделок
    return revenue[k-1]

prices = [3, 2, 6, 5, 0, 3]
k = 2
print(bookmaker(prices, k))
```
