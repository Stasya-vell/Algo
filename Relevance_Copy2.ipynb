```python
class Minheap_k:
  def __init__(self, rel, k):
    self.heap = [] # куча
    self.rel = rel # список релевантностей
    self.k = k     # размер кучи

  def parent(self, i): # O(1)
    # индекс родителя
    return i // 2

  def left_child(self, i): # O(1)
    # индекс левого ребенка
    return i * 2

  def right_child(self, i): # O(1)
    # индекс правого ребенка
    return i * 2 + 1

  def add(self, ind): # O(k)
    # Добавляем элемент в конец и проверяем балансировку если надо поднимаем элемент.
    self.heap.append(ind)
    i = len(self.heap) - 1
    p = self.parent(i)
    while i > 0 and self.rel[ind-1] < self.rel[self.heap[p]-1]:
      self.heap[i], self.heap[p] = self.heap[p], self.heap[i]
      i = p
      p = self.parent(i)

  def del_min(self): # O(k)
    # Удаляем минимальный нулевой элемент заменяя ее последним
    if self.heap:
      self.heap[0] = self.heap.pop()
      self._del_min(0)

  def _del_min(self, i): # O(k)
    # Проводим балансировку после удаления минимума.
    # Если нарушаются правила кучи меняем элементы и балансируем дальше
    x = i
    left = self.left_child(i)
    right = self.right_child(i)
    if left < self.k-1 and self.rel[self.heap[left]-1] < self.rel[self.heap[i]-1]:
      x = left
    if right < self.k-1 and self.rel[self.heap[right]-1] < self.rel[self.heap[i]-1]:
      x = right

    if i != x:
      self.heap[i], self.heap[x] = self.heap[x], self.heap[i]
      self.sift_down(x)

  def min(self): # O(1)
    # возврат минимального элемента
    return self.heap[0]

  def print(self): # O(k log k)
    # возврать отсортированной кучи
    print(*sorted(self.heap, key = lambda i: -rel[i-1]))
```

```python
n = int(input())
args = list(map(int, input().split()))
d = int(input())
fiatures = [list(map(int, input().split())) for i in range(d)]

# Создаем список всех релевантностей документов O(nd), O(d)
rel = [sum([args[j] * fiatures[i][j] for j in range(len(args))]) for i in range(d)]

q = int(input())
for _ in range(q): # O(q), O(1)
    question = list(map(int, input().split()))
    if question[0] == 2:                         # Изменяем релевантность заданного элемента
        i, j, v = question[1:]
        rel[i-1] += args[j-1] * (v - fiatures[i-1][j-1])
    elif question[0] == 1: # O(d k + k log k), O(k)
        k = question[1]
        most = Minheap_k(rel, k)                 # Создаем кучу из k элементов (индексов списка релевантности)
        for i in range(k):
          most.add(i+1)
        for i in range(k, d):                    # Если следующий элемент меньше минимума то пропускаем его, если больше то удаляем минимум и вставляем его индекс в кучу
          if rel[most.min()] < rel[i]:
            most.del_min()
            most.add(i+1)
        most.print()                             # Печатаем список индексов максимальных элементов
```
