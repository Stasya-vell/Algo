```python
class Node:
    def __init__(self, value=None, left=None, right=None, color='red', parent=None):
        self.value = value
        self.left = left
        self.right = right
        self.color = color
        self.parent = parent
```

```python
class RBtree: #O(n log n),  #O(n)
    def __init__(self):
        self.root = None

    def append(self, value): #O(log n)
        '''Добавление элемента. Проверяем наличие корневого элемента.
        Находим пустой лист удовлетворяющий условиям добавления элемента.
        Добавляем значение в дерево и проверяем балансировку.'''
        if self.root is None:
            self.root = Node(value, color='black')
            self.root.right = Node(color='black', parent = self.root)
            self.root.left = Node(color='black', parent = self.root)
            return
        curnode = self.root
        while curnode:
            if value < curnode.value:
                if curnode.left.value:
                    curnode = curnode.left
                else:
                    curnode.left = Node(value, color = 'red', parent = curnode)
                    curnode.left.right = Node(color='black', parent = curnode.left)
                    curnode.left.left = Node(color='black', parent = curnode.left)
                    if curnode.color == curnode.left.color:
                        self.balance(curnode.left)
                    return
            else:
                if curnode.right.value:
                    curnode = curnode.right
                else:
                    curnode.right = Node(value, color = 'red', parent = curnode)
                    curnode.right.right = Node(color='black', parent = curnode.right)
                    curnode.right.left = Node(color='black', parent = curnode.right)
                    if curnode.color == curnode.right.color:
                        self.balance(curnode.right)
                    return

    def find(self, value): #O(log n)
        '''Ищем элемент'''
        curnode = self.root
        while curnode.value:
            if value == curnode.value:
                return node
            if value < curnode.value:
                curnode = curnode.left 
            else:
                curnode = curnode.right
        return

    def height(self, node): #O(n)
        if node is None:
            return 0
        return max(self.height(node.left), self.height(node.right)) + int(node.color=='black')

    def rotation(self, node, direction):  #O(1)
        '''Поворачинаем дерево. Менем родителей.
        Подставлем повернутую часть дерева в основную.'''
        if direction == 'left':
            curnode = node.right
            node.right = curnode.left
            curnode.left.parent = node
            curnode.left = node
        else:
            curnode = node.left
            node.left = curnode.right
            curnode.right.parent = node
            curnode.right = node
        curnode.parent = node.parent
        if node.parent is None:
            self.root = curnode
        elif node.parent.left is node:
            node.parent.left = curnode
        else:
            node.parent.right = curnode
        node.parent = curnode
        return

    def balance(self, x): #O(1) - O(log n)
        '''Обзываем элементы. Рассматриваем три класических случая.
        Поворачиваем и перекрашиваем деревья'''
        p = x.parent
        gp = p.parent
        if p is gp.left:
            u = gp.right
            if u.color == 'red':                #дядя красный
                p.color = u.color = 'black'
                gp.color = 'red'
                if gp.parent is None:
                    gp.color = 'black'
                elif gp.parent.color == 'red':
                    self.balance(gp)
                return
            if x is p.right:                   #дядя чёрный папа и дед в разных сторонах
                self.rotation(p, 'left')
                p = x
            #дядя чёрный папа и дед в одной стороне
            p.color = 'black'
            gp.color = 'red'
            self.rotation(gp, 'right')
        else:
            u = gp.left
            if u.color == 'red':                #дядя красный
                p.color = u.color = 'black'
                gp.color = 'red'
                if gp.parent is None:
                    gp.color = 'black'
                elif gp.parent.color == 'red':
                    self.balance(gp)
                return
            if x is p.left:                   #дядя чёрный папа и дед в разных сторонах
                self.rotation(p, 'right')
                p = x
            #дядя чёрный папа и дед в одной стороне
            p.color = 'black'
            gp.color = 'red'
            self.rotation(gp, 'left')

    def print_tree(self): #O(n)
        '''Печатаем дерево.'''
        if self.root is None:
            return
        spisok = [self.root]
        while len(spisok):
            curnode = spisok[-1]
            print(curnode.value, curnode.color)
            spisok.pop(len(spisok)-1)
            if curnode.left.value:
                spisok.append(curnode.left)
                print(curnode.left.value, curnode.left.color)
            if curnode.right.value:
                spisok.append(curnode.right)
                print(curnode.right.value, curnode.right.color)
            print('-----------')
```
